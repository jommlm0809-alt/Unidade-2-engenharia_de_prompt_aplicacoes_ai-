```python
import random

def jogo_da_forca_unificado():
    """Função principal que orquestra o jogo da forca com toda a lógica unificada."""

    def escolher_palavra():
        """Escolhe uma palavra aleatoriamente de uma lista predefinida."""
        palavras = ['python', 'programacao', 'desenvolvimento', 'computador', 'algoritmo', 'modular', 'funcao']
        return random.choice(palavras).upper()

    def exibir_palavra(palavra, letras_adivinhadas):
        """Exibe a palavra com as letras adivinhadas reveladas e as não adivinhadas como sublinhados."""
        exibicao = ''
        for letra in palavra:
            if letra in letras_adivinhadas:
                exibicao += letra + ' '
            else:
                exibicao += '_ '
        return exibicao.strip()

    palavra_secreta = escolher_palavra()
    letras_adivinhadas = []
    tentativas_restantes = 6 # Número de partes do boneco da forca
    letras_erradas = []

    print("Bem-vindo ao Jogo da Forca (Versão Unificada)!")
    print("Tente adivinhar a palavra secreta.")

    while tentativas_restantes > 0:
        print(f"\nPalavra: {exibir_palavra(palavra_secreta, letras_adivinhadas)}")
        print(f"Tentativas restantes: {tentativas_restantes}")
        print(f"Letras erradas: {', '.join(letras_erradas)}")

        palpite = input("Adivinhe uma letra: ").strip().upper()

        if not palpite.isalpha() or len(palpite) != 1:
            print("Por favor, digite apenas uma letra válida.")
            continue

        if palpite in letras_adivinhadas or palpite in letras_erradas:
            print(f"Você já tentou a letra '{palpite}'. Tente outra.")
            continue

        # Lógica de verificação de letra (anteriormente na função verificar_letra)
        if palpite in palavra_secreta:
            print(f"Boa! A letra '{palpite}' está na palavra.")
            letras_adivinhadas.append(palpite)
        else:
            print(f"Ops! A letra '{palpite}' não está na palavra.")
            tentativas_restantes -= 1
            letras_erradas.append(palpite)

        if '_' not in exibir_palavra(palavra_secreta, letras_adivinhadas):
            print(f"\nParabéns! Você adivinhou a palavra: {palavra_secreta}")
            break

    if tentativas_restantes == 0:
        print("\nGAME OVER! Você ficou sem tentativas.")
        print(f"A palavra secreta era: {palavra_secreta}")

# Para iniciar o jogo, chame a função principal unificada:
while True:
    jogar_novamente = input("Deseja jogar o Jogo da Forca? (SIM/NAO): ").strip().upper()
    if jogar_novamente == 'SIM':
        jogo_da_forca_unificado()
    elif jogar_novamente == 'NAO':
        print("Obrigado por jogar! Até a próxima!")
        break
    else:
        print("Resposta inválida. Por favor, digite 'SIM' ou 'NAO'.")
```
